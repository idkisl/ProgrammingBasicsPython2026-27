## 1. Зачем нужен цикл?

Цикл используется, когда одно и то же действие нужно выполнить несколько раз.

Вместо:
```python
print("Привет!")
print("Привет!")
print("Привет!")
```

можно написать:
```python
for i in range(3):
    print("Привет!")
```

Переменная `i` называется **переменной цикла**. На каждой итерации она получает очередное значение.


### Задача 1
Выведите фразу `"Всем привет!"` 10 раз.


````python
for i in range(10):
    print("Всем привет!")
````

### Задача 2
Выведите на экран значения переменной цикла при выполнении `for i in range(5)`.


````python
for i in range(5):
    print(i)
````

## 2. Функция `range()`

Основные формы:

```python
range(stop)
range(start, stop)
range(start, stop, step)
```

Важно: значение `stop` **не входит** в диапазон.

Примеры:
```python
range(5)          # 0, 1, 2, 3, 4
range(3, 8)       # 3, 4, 5, 6, 7
range(2, 11, 2)   # 2, 4, 6, 8, 10
range(10, 0, -1)  # 10, 9, ..., 1
```


### Задача 3

````python
for i in range(5, 21):
    print(i)
````

### Задача 4

````python
for i in range(5, 26, 2):
    print(i)
````

### Задача 5

````python
for i in range(20, 0, -1):
    print(i)
````

## 5. Счётчик

Если нужно узнать **количество** элементов, удовлетворяющих условию, заводят счётчик.

```python
count = 0

for ...:
    if условие:
        count += 1
```


**Задача из СССР**

Молодой человек из СССР едет по дороге.

Он встреачет подряд N милицейских, которые измеряют его скорость.

Если скорость превышает 60 км/ч, то он нарушает и милицейский ставит дырочку на правах.

Определить количество дырочек на правах после проезда по трассе.

````python
N = int(input())

counter = 0

for i in range(N):
    velocity = int(input())
    if velocity > 60:
        print("Ставлю дырочку")
        counter += 1
print(counter)

````

## 6. Накопление суммы

Для суммы используется переменная-накопитель:

```python
total = 0

for ...:
    if условие:
        total += x
```


### Задача 10
Вводится `N`, затем `N` целых чисел. Найдите сумму чётных чисел.


````python
N = int(input())
total = 0

for i in range(N):
    x = int(input())
    if x % 2 == 0:
        total += x

print(total)
````

### Задача 11
Вводится `N`, затем `N` целых чисел. Найдите среднее арифметическое положительных чисел.

Если положительных чисел нет, выведите `"Нет положительных чисел"`.


````python
N = int(input())
total = 0
count = 0

for i in range(N):
    x = int(input())
    if x > 0:
        total += x
        count += 1

if count == 0:
    print("Нет положительных чисел")
else:
    print(total / count)
````

## 7. Флаг

Иногда нужно проверить, встретился ли хотя бы один подходящий элемент.

Для этого удобно использовать логическую переменную — **флаг**.

```python
found = False

for ...:
    if условие:
        found = True
```


### Задача 12
Вводится `N`, затем `N` целых чисел. Определите, есть ли среди них отрицательное число.


````python
N = int(input())
found = False

for i in range(N):
    x = int(input())
    if x < 0:
        found = True

if found:
    print("Есть")
else:
    print("Нет")
````

## 8. Досрочное завершение: `break`

Оператор `break` немедленно завершает ближайший цикл.

Он полезен, когда дальнейший перебор уже не нужен.


### Задача 13
Вводится `N`, затем не более `N` чисел. Если встретилось число больше 200, завершите ввод. Для остальных чисел выводите их квадраты.


````python
N = int(input())

for i in range(N):
    x = int(input())
    if x > 200:
        break
    print(x ** 2)
````

## 9. Пропуск итерации: `continue`

Оператор `continue` прекращает текущую итерацию и переходит к следующей.

Используйте его, когда некоторые элементы нужно просто пропустить.


### Задача 14
Вводится `N`, затем `N` целых чисел. Выведите квадраты только положительных чисел.


````python
N = int(input())

for i in range(N):
    x = int(input())
    if x <= 0:
        continue
    print(x ** 2)
````

## 10. Поиск максимума и минимума

Если заранее известны границы значений, можно использовать стартовое значение вне диапазона.

Но более универсальный способ — считать первое число отдельно:


````python
N = int(input())

maximum = int(input())

for i in range(N - 1):
    x = int(input())
    if x > maximum:
        maximum = x

print(maximum)
````

### Задача 15
Вводится `N >= 1`, затем `N` целых чисел. Найдите минимальное число.


````python
N = int(input())

minimum = int(input())

for i in range(N - 1):
    x = int(input())
    if x < minimum:
        minimum = x

print(minimum)
````

### Задача 16
Вводится `N`, затем `N` целых чисел. Найдите наибольшее среди чисел, кратных 3.

Если таких чисел нет, выведите `"Нет подходящих чисел"`.


````python
N = int(input())
found = False
maximum = 0

for i in range(N):
    x = int(input())
    if x % 3 == 0:
        if not found or x > maximum:
            maximum = x
            found = True

if found:
    print(maximum)
else:
    print("Нет подходящих чисел")
````

## 11. Перебор возможных значений

Цикл `for` позволяет перебрать все числа из некоторого диапазона и проверить каждое из них.

Классический пример — работа с делителями.


### Задача 17
Вводится натуральное число `x`. Выведите все его натуральные делители в порядке возрастания.


````python
x = int(input())

for d in range(1, x + 1):
    if x % d == 0:
        print(d)
````

### Задача 18

## З. Минимум и максимум

### Пример преподавателя 22
Вводится `N >= 1`, затем `N` чисел. Найдите максимальное число.

**Главная идея:** первое число читаем до цикла и считаем текущим максимумом.


````python
N = int(input())

maximum = int(input())

for i in range(N - 1):
    x = int(input())
    if x > maximum:
        maximum = x

print(maximum)
````

### Пример преподавателя 23
Вводится `N >= 1`, затем `N` чисел. Найдите минимальное число и количество раз, которое оно встретилось.

Это пример, где при нахождении нового минимума нужно не только обновить значение, но и **сбросить счётчик**.


````python
N = int(input())

minimum = int(input())
count = 1

for i in range(N - 1):
    x = int(input())

    if x < minimum:
        minimum = x
        count = 1
    elif x == minimum:
        count += 1

print(minimum)
print(count)
````

## К. Комбинированные примеры

### Пример преподавателя 28
Вводится `N`, затем `N` целых чисел.

Найдите:
- количество положительных;
- сумму положительных;
- максимальное положительное.

Если положительных чисел нет, выведите `"Нет положительных чисел"`.


````python
N = int(input())

count = 0
total = 0
found = False
maximum = 0

for i in range(N):
    x = int(input())

    if x > 0:
        count += 1
        total += x

        if not found or x > maximum:
            maximum = x
            found = True

if not found:
    print("Нет положительных чисел")
else:
    print("Количество:", count)
    print("Сумма:", total)
    print("Максимум:", maximum)
````

### Пример преподавателя 29

- меньше 100.


````python
N = int(input())
count = 0

for i in range(N):
    x = int(input())

    if x > 0 and x % 2 == 0 and x < 100:
        count += 1

print(count)
````

### Пример преподавателя 30
Вводится натуральное число `N`.

Найдите первое число от 2 до `N`, которое имеет ровно 3 делителя.

Если такого числа нет, выведите `"Нет"`.


````python
N = int(input())
found = False

for x in range(2, N + 1):
    count = 0

    for d in range(1, x + 1):
        if x % d == 0:
            count += 1

    if count == 3:
        print(x)
        found = True
        break

if not found:
    print("Нет")
````

## Вопросы после разбора примеров

После каждого примера можно менять одно условие и просить студентов устно предложить исправление:
