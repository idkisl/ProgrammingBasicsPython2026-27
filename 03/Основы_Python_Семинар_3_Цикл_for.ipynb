# Основы программирования на Python
## Семинар 3. Цикл `for`

На семинаре отрабатываются:
- `for` и `range`;
- перебор диапазонов;
- обработка `N` введённых значений;
- счётчики и накопители;
- флаги;
- `break` и `continue`;
- минимум и максимум;
- задачи на делители.

Старайтесь сначала проговорить **что именно изменяется на каждой итерации**, а затем писать код.


### Короткое повторение

Заполните устно:

1. Какие три формы `range()` вы знаете?
2. Входит ли правая граница в `range()`?
3. Как сделать цикл по убыванию?
4. Когда нужен счётчик?
5. Когда нужна переменная-сумма?
6. Для чего нужен флаг?
7. Чем отличаются `break` и `continue`?


### Задача 1
Выведите строку `"Hello everybody!"` 10 раз.


### Задача 2
Вводится целое число. Выведите это число 7 раз, каждое с новой строки.


### Задача 3
Выведите значения переменной цикла для `for i in range(8)`. Объясните, какие значения принимает `i`.


### Задача 4
Вводятся три целых числа `A`, `B`, `K`. Выведите числа от `A` до `B` включительно с шагом `K`. Считайте, что `A <= B` и `K > 0`.


### Задача 5
Выведите все числа от 30 до 5 включительно в обратном порядке.


### Задача 6
Выведите все числа от 1 до 100, кратные 7.


### Задача 7
Вводится `N`. Выведите таблицу квадратов чисел от 1 до `N`: в каждой строке число и его квадрат.


### Задача 8
Вводятся натуральные числа `A` и `N`. Выведите степени `A^0, A^1, ..., A^N`, каждую с новой строки.


### Задача 9
Вводится `N`, затем `N` целых чисел. Выведите квадрат каждого введённого числа.


### Задача 10

### Задача 11
Вводится `N`, затем `N` целых чисел. Найдите сумму чисел, кратных 5.


### Задача 12
Вводится `N`, затем `N` целых чисел. Найдите сумму нечётных чисел и их количество. Выведите оба значения.


### Задача 13
Вводится `N`, затем `N` целых чисел. Найдите среднее арифметическое чисел, кратных 3. Если таких чисел нет, выведите соответствующее сообщение.


### Задача 14
Вводится `N`, затем `N` целых чисел. Определите, есть ли среди них число из диапазона от 10 до 200 включительно.


### Задача 15
Вводится `N`, затем `N` целых чисел. Определите, есть ли среди них число, которое одновременно больше 15 и делится на 2 или на 3.


### Задача 16
Вводится `N`, затем не более `N` чисел. Выводите квадраты чисел. Если встретилось число, квадрат которого больше 400, завершите цикл, не выводя квадрат этого числа.


### Задача 17
Вводится `N`, затем `N` целых чисел. Пропускайте отрицательные числа, а для остальных выводите их удвоенное значение.


### Задача 18
Вводится `N >= 1`, затем `N` целых чисел. Найдите наибольшее число.


### Задача 19
Вводится `N >= 1`, затем `N` целых чисел. Найдите наименьшее число.


### Задача 20

### Задача 21
Вводится `N`, затем `N` целых чисел. Найдите наименьшее среди чисел, кратных 3 или 5. Если подходящих чисел нет, выведите `Нет`.


### Задача 22

### Задача 23
Вводится натуральное число `x`. Подсчитайте количество его натуральных делителей.


### Задача 24

### Задача 25
