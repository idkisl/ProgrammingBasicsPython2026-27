# Зачем нужны составные условия?

Одного сравнения часто недостаточно.

Например:

> Число должно быть **положительным И чётным**.

Нужно проверить сразу два утверждения:

```python

x > 0

x % 2 == 0


```

Как объединить их в одно условие?


# Логические операции

В Python используются три основные логические операции:

| Python | Читаем | Идея |
|---|---|---|
| `and` | И | должны выполняться оба условия |
| `or` | ИЛИ | достаточно хотя бы одного |
| `not` | НЕ | меняет истину на ложь и наоборот |

Результат логического выражения — `True` или `False`.


# `and` — логическое И

```python
A and B
```

Истинно **только тогда, когда истинны оба условия**.

| A | B | `A and B` |
|---|---|---|
| False | False | False |
| False | True | False |
| True | False | False |
| True | True | True |

Пример:

```python
x > 0 and x % 2 == 0
```

Число одновременно положительное **и** чётное.


# `and`: принадлежность промежутку

Нужно проверить:

> число `x` принадлежит отрезку `[10; 20]`

Через `and`:

```python
x >= 10 and x <= 20
```

В Python можно записать короче:

```python
10 <= x <= 20
```

Обе записи выражают одну идею: должны выполняться **обе границы**.


# `or` — логическое ИЛИ

```python
A or B
```

Истинно, если **хотя бы одно** условие истинно.

| A | B | `A or B` |
|---|---|---|
| False | False | False |
| False | True | True |
| True | False | True |
| True | True | True |

Пример:

```python
x % 3 == 0 or x % 5 == 0
```

Число делится на 3 **или** на 5.


# Важная особенность `or`

Фраза:

> делится на 3 или на 5

обычно означает **хотя бы одно из условий**.

Число `15` тоже подходит:

```python
15 % 3 == 0   # True
15 % 5 == 0   # True
```

Поэтому:

```python
True or True
```

даёт `True`.


# `not` — логическое НЕ

```python
not A
```

меняет логическое значение на противоположное.

| A | `not A` |
|---|---|
| False | True |
| True | False |

Например:

```python
not (x > 0)
```

означает:

> неверно, что `x` положительно

то есть:

```python
x <= 0
```


# Несколько операций в одном условии

Можно объединять несколько логических операций:

```python
x > 0 and (x % 2 == 0 or x % 3 == 0)
```

Читаем:

> `x` положительно **И** при этом делится на 2 **ИЛИ** на 3.

Скобки помогают явно показать смысл условия.


# Приоритет операций

Порядок выполнения:

1. сравнения: `<`, `>`, `<=`, `>=`, `==`, `!=`;
2. `not`;
3. `and`;
4. `or`.

Поэтому:

```python
A or B and C
```

равносильно:

```python
A or (B and C)
```

### Правило
Если условие сложно читать — **ставьте скобки**.


````python
print(1 > 2 or 3 < 4 and 5 != 5 or not 6 < 7 and 7 > 6 == True)
````

````python
print((1 > 2) or ((3 < 4) and (5 != 5)) or ((not (6 < 7)) and ((7 > 6) == True)))
````

````python
print((1 > 2) == (3 < 6)) # False == True
````

# Перевод условия задачи в код

Условие:

> Число положительное и двузначное или равно нулю.

Разбиваем на части:

```text
A: x > 0
B: 10 <= x <= 99
C: x == 0
```

Логическая схема:

```text
(A AND B) OR C
```

Python:

```python
(x > 0 and 10 <= x <= 99) or x == 0
```


# Вложенный условный оператор

Внутри одного `if` может находиться другой `if`.

```python
if условие_1:
    if условие_2:
        действие
```

Второе условие проверяется **только тогда**, когда первое истинно.

Это позволяет строить последовательность проверок.


# Пример вложенного `if`

Сначала проверяем логин.

Только если логин правильный — проверяем пароль.

```python
if login == "admin":
    password = input()
    if password == "1234":
        print("Доступ разрешён")
    else:
        print("Неверный пароль")
else:
    print("Неизвестный пользователь")
```

Вторая проверка имеет смысл только после успешной первой.


````python
if login == "admin" and password == "1234":
    print("Доступ разрешён")
else:
    print("Доступ не разрешён")
````

# Отступы определяют структуру

```python
if A:
    if B:
        print("1")
    else:
        print("2")
else:
    print("3")
```

`else` относится к `if`, находящемуся **на том же уровне отступа**.

В Python отступ — не оформление, а часть синтаксиса программы.


# Вложенный `if` или `and`?

### Вложенный вариант

```python
if x > 0:
    if x % 2 == 0:
        print("Подходит")
```

### Составное условие

```python
if x > 0 and x % 2 == 0:
    print("Подходит")
```

Если требуется просто проверить несколько признаков одновременно, составное условие обычно короче.

Если после первой проверки появляются **разные сценарии**, вложенный `if` часто удобнее.


# Отрицание составного условия

```python
not (x > 0 and x < 10)
```

означает:

> неверно, что `x` находится внутри `(0; 10)`.

Эквивалентно:

```python
x <= 0 or x >= 10
```

Законы де Моргана:

```text
NOT (A AND B) = (NOT A) OR (NOT B)
NOT (A OR B)  = (NOT A) AND (NOT B)
```


````python
a = 3
if a == 5 or 2:
    print("Либо 5, либо 2")
else:
    print("Иное")
````

Вводятся длины сторон `a`, `b`, `c`. Сначала определите, **существует ли треугольник**. Если существует, определите его тип: равносторонний, равнобедренный или разносторонний.


````python
a = int(input())
b = int(input())
c = int(input())

if max(a, b, c) >= (a + b + c - max(a, b, c)):
    print("Треугольника не существует")
else:
    if a == b and b == c:
        print("равносторонний")
    elif (a == b or b == c or c == a):
        print("равнобедренный")
    else:
        print("разносторонний")
````
