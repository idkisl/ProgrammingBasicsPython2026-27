# 1. `while` как другой способ записать знакомый цикл

До этого мы использовали `for`, когда заранее известно, сколько раз нужно повторить действие.

Например:

```python
for i in range(5):
    print(i)
```

То же самое можно записать через `while`.


````python
i = 0

while i < 5:
    print(i)
    i += 1
````

У такого цикла есть три важные части:

```python
i = 0          # 1. начальное значение

while i < 5:   # 2. условие продолжения
    print(i)
    i += 1     # 3. изменение переменной
```

Пока условие после `while` истинно, тело цикла выполняется.


## Пример 1. Числа от 1 до 10

Через `for`:


````python
for i in range(1, 11):
    print(i)
````

Через `while`:

````python
i = 1

while i <= 10:
    print(i)
    i += 1
````

## Пример 2. Числа от 10 до 1

Через `for`:


````python
for i in range(10, 0, -1):
    print(i)
````

Через `while`:

````python
i = 10

while i >= 1:
    print(i)
    i -= 1
````

## Пример 3. Чётные числа от 2 до 20

````python
i = 2

while i <= 20:
    print(i)
    i += 2
````

## Пример 4. Квадраты чисел от 1 до `N`

````python
N = int(input())

i = 1
while i <= N:
    print(i ** 2)
    i += 1
````

## Пример 5. Сумма чисел от 1 до `N`

````python
N = int(input())

i = 1
total = 0

while i <= N:
    total += i
    i += 1

print(total)
````

## Пример 6. Количество чисел от 1 до `N`, кратных 3

````python
N = int(input())

i = 1
counter = 0

while i <= N:
    if i % 3 == 0:
        counter += 1
    i += 1

print(counter)
````

# 2. Циклы, которые уже неудобно записывать через `for`

`while` особенно полезен, когда количество повторений заранее неизвестно.

Например:

> Вывести все натуральные числа, квадрат которых меньше `N`.


````python
N = int(input())

a = 1
while a ** 2 < N:
    print(a)
    a += 1
````

## Степени двойки, меньшие `N`

Последовательность меняется не на `+1`, а умножением на 2.


````python
N = int(input())

a = 1
while a < N:
    print(a)
    a *= 2
````

# 3. Ввод до стоп-значения

Пусть числа вводятся до тех пор, пока пользователь не введёт `0`.

Мы **не знаем**, после какого по счёту числа появится `0`.  
Поэтому `while` здесь естественнее `for`.


````python
num = int(input())

while num != 0:
    print(num)
    num = int(input())
````

Обратите внимание:

```python
num = int(input())      # первый ввод

while num != 0:         # проверка
    ...
    num = int(input())  # следующий ввод
```

Если забыть последний `input()`, значение `num` не изменится и цикл может стать бесконечным.


# 4. Условие завершения и условие продолжения

Это одна из самых важных идей занятия.

В условии задачи часто написано **когда цикл заканчивается**.

Но после `while` мы записываем **когда цикл продолжается**.

### Пример

> Ввод заканчивается, когда пользователь введёт `-1`.

Условие завершения:

```python
num == -1
```

Условие продолжения — противоположное:

```python
num != -1
```


````python
num = int(input())

while num != -1:
    print(num)
    num = int(input())
````

## Таблица: завершение → продолжение

| Условие завершения | Условие продолжения |
|---|---|
| `x == 0` | `x != 0` |
| `x == -1` | `x != -1` |
| `x > 100` | `x <= 100` |
| `x < 0` | `x >= 0` |
| `x % 2 == 0` | `x % 2 != 0` |

Сначала полезно выписать словами **условие завершения**, а затем заменить его на противоположное.


# 5. Несколько условий завершения

Рассмотрим важный пример:

> Ввод заканчивается, когда пользователь введёт `-1` **ИЛИ** `13`.

### Условие завершения

```python
num == -1 or num == 13
```

Но цикл должен продолжаться, пока **не произошло ни первое, ни второе**:

```python
num != -1 and num != 13
```


````python
num = int(input())

while num != -1 and num != 13:
    print(num)
    num = int(input())
````

### Очень важное наблюдение

При переходе от условия завершения к условию продолжения:

- `==` меняется на `!=`;
- `<` меняется на `>=`;
- `>` меняется на `<=`;
- **`ИЛИ` меняется на `И`;**
- **`И` меняется на `ИЛИ`.**

Это следствие логического отрицания условия.


## Почему нельзя написать `or`?

Неправильно:

```python
while num != -1 or num != 13:
```

Если `num == -1`, то:

- `num != -1` → `False`;
- `num != 13` → `True`.

Получается:

```python
False or True
```

то есть `True`.

Цикл продолжится.

Поэтому правильное условие:

```python
while num != -1 and num != 13:
```


## Пример 1. Заканчиваем при `0` ИЛИ `100`


````python
num = int(input())

while num != 0 and num != 100:
    print(num)
    num = int(input())
````

## Пример 2. Заканчиваем, если число отрицательное ИЛИ больше 100

Условие завершения:

```python
x < 0 or x > 100
```

Условие продолжения:

```python
x >= 0 and x <= 100
```

В Python можно записать ещё короче:

```python
0 <= x <= 100
```


````python
x = int(input())

while 0 <= x <= 100:
    print(x)
    x = int(input())
````

## Пример 3. Заканчиваем при чётном числе ИЛИ числе 15

Условие завершения:

```python
x % 2 == 0 or x == 15
```

Условие продолжения:

```python
x % 2 != 0 and x != 15
```


````python
x = int(input())

while x % 2 != 0 and x != 15:
    print(x)
    x = int(input())
````

# 6. А если завершение задано через И?

Например:

> Закончить ввод, когда число **положительное И чётное**.

Условие завершения:

```python
x > 0 and x % 2 == 0
```

Условие продолжения:

```python
x <= 0 or x % 2 != 0
```


````python
x = int(input())

while x <= 0 or x % 2 != 0:
    print(x)
    x = int(input())
````

# 7. Задачи со стоп-значением

## Пример. До первого чётного числа

Вводятся числа, пока не будет введено чётное число.  
Для всех введённых до него чисел вывести квадрат.


````python
num = int(input())

while num % 2 != 0:
    print(num ** 2)
    num = int(input())
````

## Пример. До числа, оканчивающегося на 7

Последнюю цифру удобно получать через `% 10`.

Для отрицательных чисел используем:

```python
abs(num) % 10
```


````python
num = int(input())

while abs(num) % 10 != 7:
    print(num * 2)
    num = int(input())
````

````python
print(-17 % 10)
print(-17 // 10)
a = -17
print(a, a // 10 * 10 + a % 10)


````

## Пример. Фильтрация внутри `while`

Вводятся числа до первого числа, кратного 3.  
Вывести квадраты только тех чисел, которые оканчиваются на 1.


````python
num = int(input())

while num % 3 != 0:
    if abs(num) % 10 == 1:
        print(num ** 2)
    num = int(input())
````

# 8. Счётчик

Вводятся числа до первого числа, кратного 6.  
Посчитать количество введённых до него чисел.


````python
counter = 0
num = int(input())

while num % 6 != 0:
    counter += 1
    num = int(input())

print(counter)
````

# 9. Сумма

Вводятся числа до `-15`.  
Посчитать сумму.


````python
total = 0
num = int(input())

while num != -15:
    total += num
    num = int(input())

print(total)
````

# 10. Среднее арифметическое

Вводятся числа до первого числа, оканчивающегося на 9.

Для среднего нужны одновременно:
- сумма;
- количество элементов.


````python
counter = 0
total = 0
num = int(input())

while abs(num) % 10 != 9:
    counter += 1
    total += num
    num = int(input())

if counter > 0:
    print(total / counter)
````

# 11. Строки

Стоп-значением может быть пустая строка.


````python
line = input()
counter = 0

while line != "":
    counter += 1
    line = input()

print(counter)
````

## Строки с нужной подстрокой

````python
line = input()
counter = 0

while line != "":
    if "караганда" in line:
        counter += 1
    line = input()

print(counter)
````

# 12. Минимум и максимум

Если известен диапазон, можно использовать специальное начальное значение.

Например, числа находятся от `-1000` до `1000`.  
Найдём максимум среди кратных 5.


````python
mx = -1001
num = int(input())

while num != 0:
    if num % 5 == 0 and num > mx:
        mx = num
    num = int(input())

print(mx)
````

````python
# Вывести максимум среди введенных натуральных

num = int(input())
mx = num
while num != 0:
    if num > mx:
       mx = num 
    num = int(input())
````

# 13. Более алгоритмический пример

Разложение числа на простые множители.


````python
N = int(input())
d = 2

while N > 1:
    if N % d == 0:
        print(d)
        N //= d
    else:
        d += 1
````

# 14. Типичные ошибки

1. Переменная из условия `while` не изменяется.
2. Внутри цикла забыли новый `input()`.
3. Стоп-значение случайно включили в сумму или счётчик.
4. Для завершения по `A ИЛИ B` ошибочно написали продолжение через `or`.
5. Для завершения по `A И B` забыли, что в условии продолжения нужен `or`.
6. Неправильно выбрали начальное значение минимума или максимума.
7. При вычислении среднего не проверили, что количество элементов не равно нулю.
