**Задание 17.**

Напишите функцию swap(), принимающую на вход список из двух чисел.

Функция должна вернуть список, в котором элементы поменены местами.

```python
def swap(lst):
    new_lst = [lst[1], lst[0]]
    return new_lst


numbers = [1, 2]
new_numbers = swap(numbers)
print(new_numbers)
print(numbers)
```

```python
def f(lst):
    lst.append(4)


numbers = [1, 2, 3]
print(numbers)
f(numbers)
print(numbers)
```

**Задание 18.**

Напишите функцию swap(), принимающую на вход список из двух чисел.

Функция должна поменять местами элементы списка.

При этом функция ничего не возвращает.

```python
def swap(lst):
    lst.append(lst[0])
    lst.pop(0)

numbers = [1, 2]
swap(numbers)
print(numbers)
```
