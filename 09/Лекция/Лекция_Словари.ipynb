## 1. Что такое словарь?

**Словарь** (`dict`) — изменяемая коллекция, в которой данные хранятся в виде пар **ключ: значение**.

Например, удобно хранить средний вес фруктов: ключ — название фрукта, значение — вес в граммах.

- **Ключ** должен быть уникальным в пределах одного словаря.
- **Значение** может повторяться и быть числом, строкой, списком или даже другим словарём.
- Элементы словаря сохраняют **порядок добавления** (Python 3.7+), но обращаются к ним обычно **по ключу**, а не по индексу.
- Ключ должен иметь хешируемый тип (например, `str`, `int`, `tuple` из хешируемых элементов). Список нельзя использовать в качестве ключа.

Общий вид: `словарь = {ключ1: значение1, ключ2: значение2}`.

````python
fruits = {"яблоко": 180, "банан": 140, "мандарин": 125}





print(fruits)
print("Вес яблока:", fruits["яблоко"], "г")
````

### Чем словарь отличается от списка?

В списке мы ищем элемент по **номеру** (`items[0]`), а в словаре — по **ключу** (`prices["хлеб"]`). Это удобно, когда каждому объекту соответствует характеристика.

````python
weights_list = [180, 140, 125]
weights_dict = {"яблоко": 180, "банан": 140, "мандарин": 125}
print(weights_list[0])
print(weights_dict["яблоко"])
print(weights_dict[0])
````

## 2. Создание нового и пустого словаря

Словарь можно создать сразу с элементами, пустым — с помощью `{}` или `dict()`. Функция `dict()` также позволяет собрать словарь из последовательности пар.

````python
ages = {"Анна": 16, "Борис": 17}
empty_a = {}
empty_b = dict()


print("С возрастами:", ages)
print("Пустой 1:", empty_a)
print("Пустой 2:", empty_b)
print("Тип:", type(ages))
````

### Как устроены ключи?

Ключи не повторяются: при повторной записи значение перезаписывается. Типы ключей могут различаться, но в обычных задачах лучше придерживаться одной понятной схемы ключей.

````python
example = {"кошка": 5, "кошка": 4, 10: "десять"}
print(example)
print("Количество разных ключей:", len(example))
````

## 4. Получение значения по ключу

- `d[key]` возвращает значение, **но вызовет `KeyError`**, если ключа нет.
- `key in d` проверяет **наличие ключа**, а не значения.
- `d.get(key)` вернёт `None`, если ключ отсутствует; `d.get(key, default)` — указанный запасной результат.

````python
fruits = {"яблоко": 180, "банан": 140}

print(fruits["банан"])
print("груша" in fruits)

print(fruits.get("груша")
print(fruits.get("груша", "Фрукт не найден"))
````

## 5. Добавление и изменение элементов

Одна и та же запись `d[key] = value` выполняет две разные операции:

- если ключа ещё нет — **добавляет** новую пару;
- если ключ уже есть — **заменяет** его значение.

В словаре можно хранить не только числа и строки, но и списки.

````python
fruits = {"яблоко": 180}

fruits["банан"] = 140          # добавили
fruits["яблоко"] = 175        # изменили
fruits["мандарин"] = 125      # добавили

print(fruits)
````

````python
measurements = {}
measurements["яблоко"] = [170, 180]
measurements["яблоко"].append(190)
measurements["банан"] = [130]


print(measurements)

````

## 6. Удаление элементов словаря

- `del d[key]` удаляет элемент по ключу; отсутствующий ключ вызовет ошибку.
- `d.pop(key)` удаляет элемент и **возвращает** его значение. Можно передать запасной результат: `d.pop(key, default)`.
- `d.popitem()` удаляет и возвращает **последнюю добавленную пару**.
- `d.clear()` очищает весь словарь.

````python
fruits = {"яблоко": 180, "банан": 140, "груша": 160, "киви": 90}
del fruits["груша"]


removed_weight = fruits.pop("банан")
print("Удалённый вес:", removed_weight)
print("Неизвестный фрукт:", fruits.pop("ананас", "нет такого ключа"))
last_pair = fruits.popitem()
print("Последняя пара:", last_pair)
print("Осталось:", fruits)
fruits.clear()
print("После clear:", fruits)
````

## 7. Циклы по словарю

При записи `for key in d` перебираются **ключи**. Если нужны только значения — `d.values()`. Если одновременно нужны ключи и значения — `d.items()`.

Важно: во время обычного перебора словаря **не следует добавлять или удалять ключи**. Если это необходимо, можно перебирать заранее созданный список ключей `list(d)`.

````python
fruits = {"яблоко": 180, "банан": 140, "мандарин": 125}
print("КЛЮЧИ:")
for fruit in fruits:
    print(fruit)



    
print("ЗНАЧЕНИЯ:")
for weight in fruits.values():
    print(weight)


    
print("ПАРЫ:")
for fruit, weight in fruits.items():
    print(fruit, weight)
````

### Подсчёт и накопление результатов

Для задач на частоты и суммы полезен приём `d[key] = d.get(key, 0) + value`. Если ключа не было, начинаем считать с нуля.

````python
votes = [("Иван", 3), ("Ольга", 4), ("Иван", 2), ("Ольга", 1), ("Иван", 5)]
totals = {}
for name, number in votes:
    totals[name] = totals.get(name, 0) + number
print(totals)
````

## 8. Другие полезные методы словаря

| Метод | Назначение |
|---|---|
| `d.keys()` | Представление всех ключей |
| `d.values()` | Представление всех значений |
| `d.items()` | Представление пар `(ключ, значение)` |
| `d.get(key, default)` | Безопасное получение значения |
| `d.update(other)` | Добавление и обновление пар из другого словаря |
| `d.setdefault(key, default)` | Получить значение, а если ключа нет — записать `default` |
| `d.copy()` | Создать **поверхностную** копию словаря |
| `d.pop(key, default)` | Извлечь значение с удалением ключа |
| `d.popitem()` | Удалить последнюю добавленную пару |
| `d.clear()` | Очистить словарь |

`keys()`, `values()` и `items()` возвращают **представления (views)**, а не готовые списки; при необходимости их можно превратить в списки с `list(...)`.

````python
a = {"яблоко": 180, "банан": 140}
b = {"банан": 150, "груша": 160}
print("Ключи:", list(a.keys()))
print("Значения:", list(a.values()))
print("Пары:", list(a.items()))
a.update(b)
print("После update:", a)
a.setdefault("киви", 90)
a.setdefault("банан", 999)  # существующее значение не меняется
print("После setdefault:", a)
a_copy = a.copy()
print("Копия:", a_copy)
````

### Как быстро объединить два словаря?

Для объединения используйте `update()` или оператор `|` (Python 3.9+). Если ключ встречается в обоих словарях, значение из **правого** словаря имеет приоритет.

````python
first = {"а": 1, "б": 2}
second = {"б": 20, "в": 3}
merged = first | second
print(merged)
print("Исходный first:", first)
````

## 10. Создание словаря из вводимых данных

Во многих задачах сначала считывают `N`, затем `N` строк. Метод `.split()` разбивает строку на части, а `int()` преобразует введённый текст в число. Запустите пример: он **не требует ввода** и имитирует строки пользователя.

````python
lines = ["яблоко 180", "банан 140", "яблоко 175"]
weights = {}
for line in lines:
    fruit, weight = line.split()
    weights[fruit] = int(weight)
print(weights)  # вес яблока обновлён последней записью
````

## Задача 2. Создание таблицы из N записей

Сначала вводится целое число `N`, затем `N` строк вида `название вес`. Создайте словарь `фрукт → вес` и выведите его содержимое: каждую пару в отдельной строке, разделяя ключ и значение пробелом. Если фрукт встречается повторно, сохраните **последнее** значение.

```text

```

```text

banana 140
```

## Задача 3. Несколько измерений одного фрукта

На вход поступают `N` записей `фрукт вес`. Один фрукт может встретиться несколько раз. Сохраните **все** значения веса каждого фрукта в списке. Выведите `название` и список полученных весов.

**Ввод:**
```text
4
apple 170
banana 140
apple 180
apple 190
```
**Вывод:**
```text
apple [170, 180, 190]
banana [140]
```

````python
N = int(input())
weights = {}

for i in range(N):
    line = input()
    line = line.split(" ")
    fruit = line[0]
    w = int(line[1])

    
    if fruit in weights:
        weights[fruit].append(w)
    else:
        weights[fruit] = [w]
    
print(weights)
````

## Задача 7. Карточки людей и команды изменения

Постройте вложенный словарь людей. Введите число `M`, далее `M` строк: `имя фамилия возраст гражданство`. Ключом внешнего словаря должна быть **фамилия**, а значением — словарь с полями `name`, `surname`, `age` (целое число), `citizenship`.

Затем введите `N` команд вида `фамилия:операция:значение`. Поддержите `change_name`, `change_age`, `change_citizenship`. Считайте, что запрошенная фамилия существует и фамилии не повторяются. В конце выведите для каждого человека строку `фамилия имя возраст гражданство` в порядке добавления.

**Ввод:**
```text
2
Ivan Ivanov 18 RUS
Petr Petrov 10 ARG
3
Ivanov:change_name:Vanya
Petrov:change_age:17
Ivanov:change_citizenship:RUSSIA
```
**Вывод:**
```text
Ivanov Vanya 18 RUSSIA
Petrov Petr 17 ARG
```

````python
people = {}
M = int(input())
for i in range(M):
    line = input().split(" ")
    name = line[0]
    surname = line[1]
    age = int(line[2])
    cit = line[3]

    people[surname] = {"name": name, "age": age, "citizenship": cit}

N = int(input())
for i in range(N):
    line = input()
    line = line.split(":")
    surname = line[0]
    c = line[1]
    new_value = line[2]

    if c == "change_name":
        people[surname]["name"] = new_value
    elif c == "change_age":
        people[surname]["age"] = int(new_value)
    elif c == "change_citizenship":
        people[surname]["citizenship"] = new_value
print(people)
````
