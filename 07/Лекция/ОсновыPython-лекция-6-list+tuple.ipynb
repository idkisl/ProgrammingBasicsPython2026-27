# 1. Сравним знакомые и новые типы

| Тип | Порядок | Индексы | Повторы | Можно менять |
|---|---:|---:|---:|---:|
| `str` | да | да | да | нет |
| `set` | нет | нет | нет | да |
| `list` | да | да | да | да |
| `tuple` | да | да | да | нет |

Получается:

- `list` похож на строку, но его элементы можно изменять;
- `tuple` похож на строку тем, что он неизменяем;
- `set` нужен прежде всего для уникальности;
- `list` нужен, когда важны **порядок, повторы и позиции элементов**.


# 2. Список `list`

Список — **упорядоченная изменяемая последовательность**.

Запустите ячейку и сравните список с множеством.


````python
numbers_list = [10, 20, 30, 20]
numbers_set = {10, 20, 30, 20}

print("list:", numbers_list)
print("set: ", numbers_set)
````

В списке:

- сохраняется порядок;
- сохраняются повторы;
- к элементам можно обращаться по индексам.

Поэтому список подходит для истории измерений, оценок, очереди, маршрута, последовательности команд и т. п.


# 3. Индексы: связь со строками

Со строками мы уже использовали индексы. Со списками механизм тот же.


````python
text = "python"
numbers = [10, 20, 30, 40]

print("Строка:")
print(text[0])
print(text[-1])

print("\nСписок:")
print(numbers[0])
print(numbers[-1])
````

Индексация начинается с нуля:

```text
элемент:  10   20   30   40
индекс:    0    1    2    3
```

Главное отличие: список изменяем, строка — нет.


````python
numbers = [10, 20, 30]
numbers[0] = 100

print(numbers)

# Строку так изменить нельзя:
text = "python"
text[0] = "P"
````

# 6. Создание и заполнение списка

Список можно создать сразу с элементами или пустым.

# 7. Основные методы `list`

У `str` и `set` уже были методы. У списка они тоже есть.

| Выражение | Что делает |
|---|---|
| `len(numbers)` | число элементов |
| `numbers.append(x)` | добавить `x` в конец |
| `numbers.count(x)` | число вхождений `x` |
| `numbers.index(x)` | индекс первого `x` |
| `numbers.remove(x)` | удалить первое `x` |
| `numbers.pop()` | удалить последний элемент |
| `x in numbers` | проверить наличие |


````python
numbers = [5, 8, 5, 2]

print("Список:", numbers)
print("len =", len(numbers))
print("count(5) =", numbers.count(5))
print("index(8) =", numbers.index(8))
print("8 in numbers =", 8 in numbers)
````

`remove()` удаляет только **первое** подходящее значение.


````python
numbers = [5, 10, 5, 20, 5]

print("До:", numbers)

numbers.remove(5)

print("После:", numbers)
````

`pop()` без аргумента удаляет последний элемент и возвращает его.


````python
numbers = [10, 20, 30]

last = numbers.pop()

print("Удалённый элемент:", last)
print("Осталось:", numbers)
````

# 8. Обход списка

Как и строку, список можно обходить по значениям.

Если нужен номер элемента, удобнее обходить индексы.


````python
numbers = [10, 20, 30]

for i in range(len(numbers)):
    print("Индекс:", i, "Значение:", numbers[i])
````

# 9. Срезы: почти как у `str`

Механизм срезов у `str` и `list` практически одинаков.


````python
text = "python"
numbers = [10, 20, 30, 40, 50]

print("text[1:4] =", text[1:4])
print("numbers[1:4] =", numbers[1:4])
````

Несколько полезных вариантов срезов:


````python
numbers = [10, 20, 30, 40, 50, 60]

print("Первые три:", numbers[:3])
print("С индекса 2:", numbers[2:])
print("Через один:", numbers[::2])
print("В обратном порядке:", numbers[::-1])
````

# 10. `list` и `set` решают разные задачи

Если важна **вся история** и порядок — нужен `list`.

Если нужны только **уникальные значения** — часто удобен `set`.


````python
names = ["Иванов", "Петров", "Иванов", "Сидоров"]

print("Все записи:", names)
print("Уникальные:", set(names))
print("Всего записей:", len(names))
print("Разных фамилий:", len(set(names)))
````

Часто они используются вместе.


````python
names = []

n = 4
for _ in range(n):
    names.append(input("Фамилия: "))

unique_names = set(names)

print("Все:", names)
print("Уникальные:", unique_names)
print("Всего:", len(names))
print("Различных:", len(unique_names))
````

# 11. `list` и `str` часто работают вместе

Список может хранить строки, а методы `str` применяются к каждому элементу.


````python
messages = ["error: file not found", "warning: low memory", "error: access denied"]

for message in messages:
    if message.startswith("error"):
        print(message)
````

Можно искать подстроку внутри каждого элемента списка.


````python
messages = [
    "error: file not found",
    "warning: low memory",
    "info: started"
]

for message in messages:
    if "memory" in message:
        print(message)
````

# 12. Преобразования между `str`, `list` и `set`

Строку можно превратить в список символов.


````python
text = "python"
letters = list(text)

print(letters)
````

Список можно превратить во множество и удалить повторы.

И строку можно сразу превратить во множество символов.

# 14. Зачем хранить данные, если можно обработать сразу?

Если требуется только один результат, список иногда не нужен.


````python
total = 0

for _ in range(3):
    total += int(input("Число: "))

print("Сумма:", total)
````

Но список нужен, если данные понадобятся **после ввода**:

- найти среднее и снова сравнить с ним элементы;
- работать с соседями;
- обращаться по номеру;
- выводить в обратном порядке;
- выполнять несколько разных анализов.


````python
numbers = [10, 30, 20, 40]

average = sum(numbers) / len(numbers)

print("Среднее:", average)
print("Элементы выше среднего:")

for x in numbers:
    if x > average:
        print(x)
````

# 15. Кортеж `tuple`

Кортеж — **упорядоченная неизменяемая последовательность**.

Он похож на `str`: есть индексы и срезы, но нельзя заменить отдельный элемент.


````python
point = (10, 20)
rgb = (255, 128, 0)
student = ("Иванов", 17)

print(point)
print(rgb)
print(student)
print("Первая координата:", point[0])
````

Попытка изменить кортеж привела бы к ошибке:


````python
point = (10, 20)

print("Кортеж:", point)

point[0] = 100
````

````python
# Вопрос на пятерку. Как создать tuple из одного элемента?

a = tuple([1]) # (1 + 0) * 2
print(a)


# (1,5 + 2) --> (1, 7)

print(1,2+7)
````

# 16. Зачем нужен `tuple`, если есть `list`?

Кортеж удобно использовать для одной логически цельной записи:

- координаты точки;
- цвет RGB;
- пара «имя — результат».


````python
point = (10, 20)
color = (255, 128, 0)
student = ("Иванов", 85)

print(point)
print(color)
print(student)
````

При этом набор таких записей удобно хранить в списке.


````python
points = [
    (1, 2),
    (5, 3),
    (-2, 4)
]

print(points)
print("Первая точка:", points[0])
print("x первой точки:", points[0][0])
````

# 17. Список кортежей

Это очень распространённая структура.


````python
students = [
    ("Иванов", 85),
    ("Петров", 92),
    ("Сидоров", 78)
]

for student in students:
    print(student)
````

Можно сразу распаковывать кортеж в несколько переменных.


````python
students = [
    ("Иванов", 85),
    ("Петров", 92),
    ("Сидоров", 78)
]

for name, score in students:
    print(name, "—", score)
````

# 18. Как выбрать тип данных

Задайте себе вопросы.

### Нужен ли порядок?

Нет → возможно, `set`.

Да → `str`, `list` или `tuple`.

### Нужны ли повторы?

Нет, нужны только уникальные элементы → `set`.

Да → `list`, `tuple` или `str`.

### Нужно ли менять отдельные элементы?

Да → `list`.

Нет → возможно, `tuple` или `str`.

### Данные — текст?

Да → `str`.

### Это несколько связанных значений одной записи?

Часто удобно использовать `tuple`.

## Задача 8. Сжатие последовательности

Вводится `N`, затем `N` целых чисел.

Создайте новый список, удалив только **соседние повторы**.

Пример:

```text
1 1 1 2 2 1 3 3
```

Результат:

```text
1 2 1 3
```

Объясните, почему использование только `set` здесь не решает задачу.


````python
lst = []
N = int(input())
for i in range(N):
    line = int(input())
    lst.append(line)

new_lst = [lst[0]]

for i in range(1, len(lst)):
    if lst[i - 1] != lst[i]:
        new_lst.append(lst[i])
print(new_lst)

````
