**Задание 11.**

Вводятся целые числа через пробел.

Найдите минимальное среди них и положите на 0-ой индекс в список.

````python
line = input()
lst = line.split()
lst = [int(elem) for elem in lst]



for i in range(len(lst)):
    if lst[0] > lst[i]:
        t = lst[i]
        lst[i] = lst[0]
        lst[0] = t
        
print(lst)
````

**Selection Sort**

````python
line = input()
lst = line.split()
lst = [int(elem) for elem in lst]


for j in range(len(lst)):
    for i in range(j, len(lst)):
        if lst[j] > lst[i]:
            t = lst[i]
            lst[i] = lst[j]
            lst[j] = t
        
print(lst)


````

**Задание 14.**

Вводится два числа N и M.

Далее следует N строчек по M элементов

Выведите эти же числа на экран.




```
4 3
-99 6 -16
3 16 56
-29 88 -1
-62 49 36
```




````python
line = input()
line = line.split(" ")
line = [int(i) for i in line]

N = line[0]
M = line[1]
print(N, M)

matrix = []


for i in range(N):
    line = input()
    line = line.split(" ")
    line = [int(i) for i in line]
    matrix.append(line)

print(matrix)
````

**Задание 15.**

Вводится два числа N и M.

Далее следует N строчек по M элементов

Выведите максимальное число среди матрицы на экран.

````python
line = input()
line = line.split(" ")
line = [int(i) for i in line]

N = line[0]
M = line[1]
print(N, M)

matrix = []


for i in range(N):
    line = input()
    line = line.split(" ")
    line = [int(i) for i in line]
    matrix.append(line)

print(matrix)

mx = matrix[0][0]
for x in range(N):
    for y in range(M):
        if matrix[x][y] > mx:
            mx = matrix[x][y]
print(mx)
````
