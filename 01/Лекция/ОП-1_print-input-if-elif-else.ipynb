# 2. Значения и типы данных

У каждого значения есть **тип**. На старте нам нужны четыре основных типа:

| Тип | Что хранит | Примеры |
|---|---|---|
| `int` | целые числа | `7`, `0`, `-42` |
| `float` | вещественные числа | `3.14`, `-0.5` |
| `str` | текст | `"Python"`, `'42'` |
| `bool` | логическое значение | `True`, `False` |

Тип объекта можно узнать функцией `type()`.


````python
print(type(42)) # int
print(type(3.14)) # float
print(type("42")) # str
print(type(True)) # bool
````

Обратите внимание: `42` и `"42"` — **не одно и то же**.  
Первое значение является числом, второе — строкой.

# 4. Арифметика в Python

Основные операции:

| Операция | Запись | Пример |
|---|---|---|
| сложение | `+` | `7 + 3` |
| вычитание | `-` | `7 - 3` |
| умножение | `*` | `7 * 3` |
| обычное деление | `/` | `7 / 3` |
| целая часть от деления | `//` | `7 // 3` |
| остаток от деления | `%` | `7 % 3` |
| степень | `**` | `7 ** 2` |

Операции `//` и `%` особенно важны для дальнейших задач с условиями.


````python
a = 17
b = 5

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a ** 2)

print(a // b)
print(a % b)
# 17 : 5 = 3 (ост. 2) = 3 + 2/5
````

### Зачем нужен остаток `%`

- последние две цифры — `n % 100`.

Например, у `347` последняя цифра равна `347 % 10`, то есть `7`.


````python
n = 347
print("Последняя цифра:", n % 10)
print("Последние две цифры:", n % 100)
````

# 7. Сравнения

Сравнение даёт логическое значение `True` или `False`.

| Оператор | Смысл |

# 9. Условный оператор `if`

До этого программа выполняла почти все команды подряд.  
`if` позволяет выполнить блок кода **только при выполнении условия**.

```python
if условие:
    команда
    команда
```

После условия ставится `:`, а тело условия записывается с отступом.


````python
temperature = 20

if temperature > 30:
    print("Сегодня жарко")
print("Проверка завершена")
````

# 10. Несколько вариантов: `if / elif / else`

Когда вариантов больше двух, удобно использовать `elif`.

```python
if условие_1:
    ...
elif условие_2:
    ...
else:
    ...
```

Python проверяет ветки **сверху вниз** и выполняет первую подходящую.


````python
score = int(input())

if score >= 90:
    print("Отлично")
elif score >= 75:
    print("Хорошо")
elif score >= 60:
    print("Зачёт")
else:
    print("Нужно доработать")
````

### Почему порядок условий важен?

Если сначала проверить `score >= 60`, то значение `95` уже попадёт в эту ветку, и до `score >= 90` программа не дойдёт.

При проверке диапазонов часто удобно идти **от более строгого условия к менее строгому**.
