## Множества (`set`)

## 1. Понятие множества: связь с математикой

**Множество в математике** — совокупность различных объектов, объединённых по некоторому признаку. Объекты называют **элементами множества**.

Например, множество чётных натуральных чисел, меньших 10:

\[
A = \{2,4,6,8\}.
\]

Запись \(4 \in A\) означает «4 принадлежит множеству A», а \(5 \notin A\) — «5 не принадлежит A».

Важные свойства математического множества:

- **Повторений нет:** \(\{1,1,2\}=\{1,2\}\).
- **Порядок не важен:** \(\{2,1,3\}=\{3,2,1\}\).
- **Пустое множество** не содержит элементов: \(\varnothing\).
- Множества можно объединять, пересекать и сравнивать.

**В Python** математической идее соответствует тип `set`. Например, `a = {2, 4, 6, 8}`. Операции с ним похожи на математические, но есть особенности: множество изменяемо, а его элементы должны быть хешируемыми.

### Где это полезно?

Когда нужно посчитать **различные** значения, проверить, встречалось ли число, найти общих участников двух групп или исключить уже известные элементы.

**Важно:** если в задаче требуется учитывать, сколько раз встречалось одно и то же число, одного множества недостаточно — оно не хранит повторы. Иногда достаточно считать ответ прямо во время ввода.

```python
numbers = {3, 1, 3, 2, 1}
print(numbers)
print(len(numbers))
```

## 2. Создание множества

- `{1, 2, 3}` — множество с элементами;
- `set()` — пустое множество (не `{}`);
- `set('мама')` — множество символов строки;
- `set(range(1, 6))` — множество чисел от 1 до 5.

Внутри множества могут быть числа и строки, но не всякие объекты: элементы должны быть хешируемыми.

```python
a = {10, 20, 10}
b = set()
c = set('мама')
d = set(range(1, 6))
print(a)
print(b)
print(c)
print(d)
```

## 3. Принадлежность и размер

`x in s` проверяет, есть ли элемент в множестве; `x not in s` — отсутствует ли. `len(s)` возвращает число различных элементов. Проверка принадлежности в среднем выполняется быстро.

```python
colors = {'красный', 'синий', 'зелёный'}
print('синий' in colors)
print('жёлтый' not in colors)
print(len(colors))
```

## 4. Изменение множества

- `add(x)` — добавить один элемент;
- `remove(x)` — удалить; ошибка, если элемента нет;
- `discard(x)` — удалить, если существует; ошибки при отсутствии нет;
- `pop()` — удалить и вернуть **произвольный** элемент (не последний);
- `clear()` — очистить.

Для добавления нескольких значений можно вызывать `add` в цикле.

```python
s = {2, 4}
s.add(6)
s.add(2)
s.discard(10)
s.remove(4)
print(s)
s.clear()
print(s)
```

## 5. Перебор множества

Цикл `for` позволяет посетить все элементы, но **не в порядке ввода**. Нельзя обращаться к элементу множества по индексу `s[0]`.

```python
s = {7, 3, 9}
for number in s:
    print(number)
```

## 6. Операции над множествами

Пусть `A` и `B` — множества.

| Операция | Оператор | Метод | Значение |
|---|---|---|---|
| Объединение | `A \| B` | `A.union(B)` | Встречаются хотя бы в одном |
| Пересечение | `A & B` | `A.intersection(B)` | Встречаются в обоих |
| Разность | `A - B` | `A.difference(B)` | Только в `A` |
| Симметрическая разность | `A ^ B` | `A.symmetric_difference(B)` | Ровно в одном |

Операции создают новые множества и не меняют исходные.

### Круги Эйлера

Закрашенная область показывает результат операции. На рисунках A и B — исходные множества.

**Объединение A ∪ B**



**Пересечение A ∩ B**



**Разность A − B**



**Симметрическая разность A △ B**



```python
a = {1, 2, 3, 4}
b = {3, 4, 5}
print(a | b)
print(a & b)
print(a - b)
print(b - a)
print(a ^ b)
```

## 7. Отношения между множествами

`A <= B` означает, что `A` — подмножество `B`; `A < B` — строгое подмножество; `A >= B` — надмножество; `A.isdisjoint(B)` — нет общих элементов. `==` сравнивает состав элементов, а не порядок.

### Геометрический смысл отношений множеств

**Подмножество:** все элементы A входят в B, поэтому круг A находится внутри круга B.

**A ⊆ B — подмножество**



**Непересекающиеся множества:** общих элементов нет, круги не имеют общей области.

**A ∩ B = ∅ — общих элементов нет**



```python
a = {2, 4}
b = {2, 4, 6}
print(a <= b)
print(a < b)
print(a == b)
print(a.isdisjoint({1, 3, 5}))
```

## 8. Построение множества с помощью цикла

Для неизвестного заранее количества значений можно создать пустое множество и добавлять элементы по мере ввода.

```python
n = int(input())
unique_numbers = set()
for _ in range(n):
    number = int(input())
    unique_numbers.add(number)
print(len(unique_numbers))
```
