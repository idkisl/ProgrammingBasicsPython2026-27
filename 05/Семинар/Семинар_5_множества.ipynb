# Семинар 5. Множества в Python (`set`)

## 1. Что такое множество?

В математике множество — совокупность различных элементов, например \(A=\{2,4,6\}\). В Python `set` хранит **уникальные** значения, а порядок элементов не фиксирован. Пустое множество создаётся через `set()`, не через `{}`.


```python
a = {2, 4, 6}
print(len(a))
a.add(4)
print(len(a))
a.add(8)
print(a)
```

## 2. Добавление, удаление, проверка

- `s.add(x)` — добавить значение.
- `x in s` — проверить принадлежность.
- `s.discard(x)` — удалить, если значение есть.
- `s.remove(x)` — удалить, но при отсутствии будет ошибка.
- `len(s)` — число элементов в множестве.


```python
seen = set()
for _ in range(4):
    x = int(input())
    seen.add(x)
print(len(seen))
print(7 in seen)
seen.discard(7)
print(seen)
```

## 3. Ввод до нуля

Вводим числа, пока не встретится ноль. При необходимости сохраняем только уникальные значения.


```python
numbers = set()
while True:
    x = int(input())
    if x == 0:
        break
    numbers.add(x)
print(len(numbers))
```

## 4. Операции над множествами

| Математика | Python | Смысл |
|---|---|---|
| \(A\cup B\) | `a | b` | Все элементы из A или B |
| \(A\cap B\) | `a & b` | Общие элементы |
| \(A\setminus B\) | `a - b` | Есть в A, но нет в B |
| \(A\triangle B\) | `a ^ b` | Есть ровно в одном множестве |

```python
a = {1, 2, 3, 4}
b = {3, 4, 5}
print(a | b)
print(a & b)
print(a - b)
print(a ^ b)
```

## 5. Подмножества и сравнение

`a <= b` — каждый элемент A содержится в B; `a < b` — строгое подмножество; `a == b` — множества совпадают; `a.isdisjoint(b)` — общих элементов нет.


```python
a = {2, 4}
b = {2, 4, 6}
print(a <= b)
print(a < b)
print(a == b)
print(a.isdisjoint({1, 3}))
```
