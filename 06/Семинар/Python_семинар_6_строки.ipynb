# Основы Python — семинар 6
## Строки


## Разминка

**Задание 1.** Вводится строка. Выведите:
1. её длину;
2. первый символ;
3. последний символ;
4. второй символ с конца.


**Задание 2.** Вводится слово. Выведите его символы в столбик.


**Задание 3.** Вводится слово. Выведите пары `индекс — символ` для каждого символа.


**Задание 4.** Вводится строка. Выведите её:
- без первого символа;
- без последнего символа;
- без первого и последнего символов;
- в обратном порядке.


## Индексы и срезы

**Задание 5. Конструктор слов.**  
Из слова `вертикаль` с помощью индексов и срезов получите:
- `тир`;
- `ветка`.

Каждый результат выведите отдельной строкой.


**Задание 6. Перестановка.**  
Вводится слово длины не менее 4 символов. Выведите новую строку, в которой:
- первые два символа перенесены в конец;
- остальные символы остаются в прежнем порядке.


**Задание 7. Зеркало.**  
Вводится слово. Выведите строку вида:

`слово|оволс`

Например, для `кот` результат: `кот|ток`.


## Подсчёт и поиск

**Задание 8.** Вводится предложение и затем один символ. Определите, сколько раз этот символ встречается в предложении.


**Задание 9.** Вводится предложение. Определите количество букв `о` без учёта регистра.


**Задание 10.** Вводится строка и символ. Выведите индекс первого вхождения символа. Если его нет, выведите `-1`.


**Задание 11. Две границы.**  
Вводится строка и символ. Если символ встречается хотя бы один раз, выведите индекс первого и последнего вхождения. Иначе выведите `Нет символа`.


## Регистр и проверки

**Задание 12.** Вводится строка. Выведите только те символы, которые записаны в верхнем регистре.


**Задание 13.** Вводится строка. Выведите индексы всех символов, являющихся цифрами.


**Задание 14. Проверка кода.**  
Вводится строка. Выведите `YES`, если она состоит только из букв и цифр, иначе `NO`.


**Задание 15. Идентификатор.**  
Вводится строка. Считайте её допустимым простым идентификатором, если:
- строка не пустая;
- первый символ — буква;
- все символы строки — буквы или цифры.

Выведите `YES` или `NO`.


## Работа с началом и концом строки

**Задание 16. Файл.**  
Вводится имя файла. Определите, заканчивается ли оно на `.py`. Регистр учитывать.


**Задание 17. Протокол.**  
Вводится адрес. Выведите `WEB`, если он начинается с `http://` или `https://`, иначе `OTHER`.


## Построение новой строки

**Задание 18. Исправление опечатки.**  
Вводится слово длины не менее 3 символов. Замените второй символ на `*`, не изменяя остальные.


**Задание 19. Цензор.**  
Вводится строка. Замените все вхождения слова `кот` на `***` и выведите результат.


**Задание 20. Смена регистра.**  
Вводится строка. Создайте новую строку, где каждая строчная буква становится заглавной, а заглавная — строчной.


## Комбинированные задачи

**Задание 21. Номер билета.**  
Вводится строка из 6 символов. Проверьте, что все 6 символов — цифры. Если нет — выведите `ERROR`.  
Если да, сравните первый и последний символ: выведите `SAME`, если они совпадают, иначе `DIFFERENT`.


**Задание 22. Никнейм.**  
Вводится никнейм. Он считается корректным, если:
- длина от 5 до 15 символов включительно;
- нет пробелов;
- первый символ — буква;
- последний символ — буква или цифра.

Выведите `OK` или `BAD`.


**Задание 23. Секретное сообщение.**  
Вводится строка. Выведите символы, стоящие на чётных индексах, затем с новой строки — символы на нечётных индексах.

Пример для `abcdef`:
`ace`
`bdf`


**Задание 24. Палиндром.**  
Вводится слово. Определите, читается ли оно одинаково слева направо и справа налево. Регистр не учитывать.


**Задание 25. Почти палиндром.**  
Вводится строка. Сначала приведите её к нижнему регистру.  
Определите, совпадает ли первый символ с последним, второй — с предпоследним и так далее.  
Решение выполните циклом, не используя готовое сравнение со строкой `[::-1]`.


## Задачи с легендой

**Задание 26. Космический позывной.**  
На станции принимают позывной пилота. Корректный позывной:
- начинается с `X`;
- заканчивается цифрой;
- не содержит пробелов;
- имеет длину не менее 6 символов.

Выведите `ACCESS`, если позывной корректен, иначе `DENIED`.


**Задание 27. Архив робота.**  
Робот сохранил строку телеметрии. Символ `!` означает аварийную отметку.

Выведите:
1. количество `!`;
2. индекс первой аварийной отметки;
3. индекс последней аварийной отметки.

Если аварийных отметок нет, вместо индексов выведите `-1` и `-1`.


**Задание 28. Шифр диспетчера.**  
Вводится сообщение. Постройте новую строку по правилу:
- символы на чётных индексах оставить как есть;
- символы на нечётных индексах заменить на `*`.

Например:  
`PYTHON` → `P*T*O*`.


**Задание 29. Турникет.**  
Система считывает строку пропуска. Пропуск верный, если:
- первые два символа — `IT`;
- последние четыре символа — цифры;
- длина строки ровно 8.

Выведите `OPEN` или `CLOSED`.


**Задание 30. Самая частая буква из двух.**  
Вводится строка, затем две буквы `a` и `b`. Без учёта регистра определите, какая из двух букв встречается чаще.  
Если количества равны, выведите `EQUAL`.
