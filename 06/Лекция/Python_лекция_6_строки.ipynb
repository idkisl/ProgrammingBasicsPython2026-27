## 3. Перебор строки циклом `for`

Если важен **сам символ**:


```python
s = "Python"

for ch in s:
    print(ch)
```

## 5. Подсчёт и поиск

Полезные строковые методы:
- `s.count(x)` — сколько раз встречается `x`;
- `s.find(x)` — позиция первого вхождения, а если ничего нет — `-1`;
- `s.rfind(x)` — позиция последнего вхождения.


```python
s = "abracadabra"

print(s.count("a"))
print(s.find("bra"))
print(s.rfind("bra"))
print(s.find("xyz"))
```

### Почему `find` удобен?
Он позволяет безопасно проверить наличие фрагмента по индексу результата.


```python
s = input("Строка: ")
fragment = input("Что ищем: ")

pos = s.find(fragment)

if pos == -1:
    print("Не найдено")
else:
    print("Первое вхождение начинается с индекса", pos)
```

Часто ввод нужно сравнивать **без учёта регистра**:

## 7. Проверки содержимого строки

Методы возвращают `True` или `False`:
- `isalpha()` — только буквы;
- `isdigit()` — только цифры;
- `isalnum()` — буквы и/или цифры;
- `islower()` — буквенные символы в нижнем регистре;
- `isupper()` — буквенные символы в верхнем регистре.


```python
s = "Python"
print(s, s.isalpha(), s.isdigit(), s.isalnum())

s = "2026"
print(s, s.isalpha(), s.isdigit(), s.isalnum())

s = "Python2026"
print(s, s.isalpha(), s.isdigit(), s.isalnum())
```

В следующем примере коллекции не создаём: просто меняйте значение `s` и перезапускайте ячейку.


```python
s = "Python2026"

print("Только буквы:", s.isalpha())
print("Только цифры:", s.isdigit())
print("Буквы/цифры:", s.isalnum())
```

```python
s = "Hello"
print(s.islower())
```

## 11. Комбинированный пример

**Задача.** Вводится строка. Нужно:
1. посчитать латинские буквы `a` без учёта регистра;
2. определить, есть ли цифра `7`;
3. вывести строку наоборот;
4. вывести индексы заглавных букв.


```python
s = input("Введите строку: ")

print("Количество a:", s.lower().count("a"))
print("Есть цифра 7:", "7" in s)
print("Наоборот:", s[::-1])

print("Индексы заглавных букв:")
for i in range(len(s)):
    if s[i].isupper():
        print(i)
```

Сначала вводится число `n`, затем `n` строк. Найдите номер строки, в которой встречается наибольшее количество различных символов. Если таких строк несколько, выведите номер первой из них.



```python
# Вводится строка. Выведите количество различных символов, встречающихся в ней.

line = input()
line_s = set(line)
print(len(line_s))
```

```python
# Вводится строка из строчных латинских букв. Выведите все различные буквы строки в алфавитном порядке.

line = input()
line_s = set(line)
for i in range(len(line_s)):
    print(min(line_s))
    line_s.remove(min(line_s))
```

```python
# Сначала вводится число n, затем n строк. Выведите символы, которые встречаются в каждой из введённых строк.

n = int(input())
line = input()
s = set(line)

for i in range(n - 1):
    line = input()
    line_s = set(line)
    s = s.intersection(line_s)
print(s)
```
